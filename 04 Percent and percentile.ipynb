```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("Titanic-Dataset.csv")
data.head()
```

```python
data.isnull().sum()
```

```python
data["Age"] = data["Age"].fillna(data["Age"].mean(),inplace=True)
```

### Percentile

```python
np.percentile(data["Age"],25)
```

```python
data_new = np.array([1,2,3,5,11,11,11,10,11,11,11])
```

```python
np.percentile(data_new,24)
```

```python
sns.boxplot(x="Age",data=data)
plt.show()
```
