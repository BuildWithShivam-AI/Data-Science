Mean = Sum of all number/ count of number

```python
data['Age'].mean()
```

```python
mean_data = np.mean(data['Age'])
```

```python
sns.histplot(x='Age',data=data,bins= [i for i in range(0,100,10)])
plt.plot([mean_data for i in range(0,500)],[i for i in range(0,500)],c= 'red',label = "Mean of Age")
plt.legend()
plt.title("Titanic pessanger of Age ")
plt.show()
```

### Median
Meddle value of the data series 
1. step 1 -> Arrange the data in assending order
2. find data is present in ever or odd number of data (like data roe is 20 or 21 )

##### Odd
eg : { 1,3,4,5,6,8,7}
- formula ==> (Middle value of data series)
* Output = 5

##### Even
eg : {1,3,4,5,6,6}
- formula ==> (middel value is {4,5})
then calculation is ([4+5]/2),  ==>

9/2 = 4.5 
* Output = 4.5


```python
np.median(data['Age'])
```

```python
Md = data['Age'].median() + 10
```

```python
sns.histplot(x='Age',data=data,bins= [i for i in range(0,100,10)])
plt.plot([mean_data for i in range(0,400)],[i for i in range(0,400)],c= 'red')
plt.plot([Md for i in range(0,400)],[i for i in range(0,400)],c= 'blue')
plt.show()
```

### Mode
Mode mean most repeted vlaue is that data set 

##### high frequency
eg : { 1,3,4,5,6,8,3,7}
- formula ==> (3 is presnet 2 time and other is 1 time )
* Output = 3



```python
data['Sex'].mode()[0]
```

```python
mode_age = data["Age"].mode()[0] + 5
mode_age
```

```python
data['Age'].value_counts()
```

```python
sns.histplot(x='Age',data=data,bins= [i for i in range(0,100,10)])
plt.plot([mean_data for i in range(0,400)],[i for i in range(0,400)],c= 'red',label ="Mean")
plt.plot([Md for i in range(0,400)],[i for i in range(0,400)],c= 'blue',label ="Median + 10")
plt.plot([mode_age for i in range(0,400)],[i for i in range(0,400)],c= 'green',label ="Mode + 5")
plt.legend()
plt.title("Working on Central Tendency")
plt.show()
```
