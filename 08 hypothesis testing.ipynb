```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as st
```

```python
old_design_data = np.array([45.2, 42.8, 38.9, 43.5, 41.0, 44.6, 40.5, 42.7, 39.8, 41.4, 44.3, 39.7, 42.1, 40.6, 43.0, 42.2, 41.5, 39.6, 44.0, 43.1, 38.7, 43.9, 42.0, 41.9, 42.8, 43.7, 41.3, 40.9, 42.5, 41.6])
new_design_data = np.array([48.5, 49.1, 50.2, 47.8, 48.7, 49.9, 48.0, 50.5, 49.8, 49.6, 48.2, 48.9, 49.7, 50.3, 49.4, 50.1, 48.6, 48.3, 49.0, 50.0, 48.4, 49.3, 49.5, 48.8, 50.6, 50.4, 48.1, 49.2, 50.7, 50.8])
```

```python
mean_mew = np.mean(new_design_data)
mean_old = np.mean(old_design_data)
```

```python
pop_std = 2.5
n = len(new_design_data)
apl = 0.05
```

```python
z_cal = (mean_mew - mean_old)/(pop_std/np.sqrt(n))
z_cal
```

```python
z_table = st.norm.ppf(1-apl)
z_table
```

```python
if z_cal > z_table:
    print("Ha is right")
else:
    print("H0 is right")
```

# One-Sample T-Test Formula

$
t = \frac{\bar{x} - \mu}{\frac{s}{\sqrt{n}}}
$

Where:

- $\bar{x}$ = Sample Mean
- $\mu$ = Population Mean
- $s$ = Sample Standard Deviation
- $n$ = Number of Samples

## Steps for Hypothesis Testing

1. State Null Hypothesis (**H₀**) and Alternative Hypothesis (**H₁**)
2. Choose Significance Level (**α**)
3. Calculate the Test Statistic (**t**)
4. Find the Critical Value or p-value
5. Draw the Conclusion

### Decision Rule

- If **|t calculated| > t critical** → Reject H₀
- Otherwise → Fail to Reject H₀

```python
observes = np.array([22,17,20,26,22,13])
expected = np.array([20,20,20,20,20,20])

alp = 0.05
chi_table = 11.07

chi_cal = sum(np.square(observes - expected)/expected)
chi_cal
```

```python
data = pd.DataFrame({"observes":observes,"expected":expected})
data.head()
```

```python
import seaborn as sns

sns.kdeplot(x="observes", data=data)
plt.plot([alp for i in range(0,1)],[i for i in range(0,1)],c="red")
plt.show()
```

```python
faces = ['1', '2', '3', '4', '5', '6']
```

```python
x = np.arange(len(faces))
width = 0.35
```

```python
plt.figure(figsize=(10,5))

plt.bar(x-width/2, observes, width, label='Observed')
plt.bar(x+width/2, expected, width, label='Expected')
plt.xticks(x, faces)
plt.xlabel("Die Face")
plt.ylabel("Frequency")
plt.title("Observed vs Expected Frequencies")
plt.legend()

plt.show()
```

```python
residuals = [(o-e)/np.sqrt(e) for o,e in zip(observes, expected)]

plt.figure(figsize=(8,4))
plt.bar(faces, residuals)

plt.axhline(0, linestyle='--')

plt.xlabel("Die Face")
plt.ylabel("Standardized Residual")
plt.title("Chi-Square Residuals")

plt.show()
```

```python
from scipy.stats import chisquare

observes = [22,17,20,26,22,13]
expected = [20,20,20,20,20,20]

chi2, p = chisquare(
    f_obs=observes,
    f_exp=expected
)

print("Chi-Square Statistic:", chi2)
print("P-value:", p)
```
