```python
#We are getting information like 
# 0 = White(Not Rating), 
# 0.1 to 2.5 = Red(Poor), 
# 2.5 to 3.5 = Orange(Average), 
# 3.5 to 4.0 = Yellow(Good), 
# 4 to 4.5 = Green(Very Good), 
# 4.5 to 5 = Dark Green(Excellent)

ratings = final_df.groupby(['Aggregate rating', 'Rating color', 'Rating text']).size().reset_index().rename(columns={0:'Rating count'})
```

6. 4.5 to 5 = Dark Green(Excellent)

```python
# Assignment: Find the top 10 Cuisines
final_df['Cuisines']
```

```python
top_10_cuisines = (
    final_df['Cuisines']
    .value_counts()
    .head(10)
)

print(top_10_cuisines)
```

```python
import matplotlib.pyplot as plt

top_10 = final_df['Cuisines'] \
    .value_counts() \
    .head(10)

plt.figure(figsize=(12,6))

top_10.plot(
    kind='bar',
    color='skyblue'
)

plt.title(
    "Top 10 Cuisines",
    fontsize=16,
    fontweight='bold'
)

plt.xlabel("Cuisine")
plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()
```

```python
# Split cuisines by comma
separate_cuisines = (
    final_df['Cuisines']
    .str.split(',')
)

# Convert nested list into single rows
separate_cuisines = (
    separate_cuisines
    .explode()
)

# Remove extra spaces
separate_cuisines = (
    separate_cuisines
    .str.strip()
)

# Count cuisines
top_cuisines = (
    separate_cuisines
    .value_counts()
)

print(top_cuisines.head(10))
```

```python
import matplotlib.pyplot as plt

top_10 = top_cuisines.head(10)

plt.figure(figsize=(12,6))

top_10.plot(
    kind='bar',
    color='orange'
)

plt.title("Top Individual Cuisines")

plt.xlabel("Cuisine")
plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()
```
