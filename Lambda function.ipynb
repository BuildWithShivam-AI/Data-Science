```python
#convert a list of integar to their correspongin square values:
number = [2,4,2,34,5]
square_value = list(map(lambda x : x**2,number))
print(square_value)
```

```python
#Filter out the even number from a list

number = [1,4,2,5,6,7,3,2,3,29]
list(filter(lambda x:x%2==0,number))
```

```python
#sort a list of string based on thire alph charater and length

words =["apple","banana","cherry","graps","banana"]
sorted(words,key=lambda x: len(x))
```
