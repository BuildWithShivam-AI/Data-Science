```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
X, y = make_regression(
    n_samples=10,
    n_features=1,
    n_informative=1,
    n_targets=1,
    noise=80,
    random_state=13
)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, y)
```

```python
# lets apply OLS

from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(X, y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color ='red')
```

```python
# Lets apply Gradient Descent assuming slope is constant m = 48.53
# and let's assume the starting value for intercept b = 100

y_pred = ((48.53 * X) + (100)).reshape(10)
```

```python
plt.scatter(X, y)

plt.plot(
    X,
    reg.predict(X),
    color='red',
    label='OLS'
)

plt.plot(
    X,
    y_pred,
    color='#00a65a',
    label='b = 0'
)

plt.legend()

plt.show()
```

```python
m = 48.53
b = 100
loss_slope = -2 * np.sum(y - m * X.ravel() - b)
loss_slope
```

```python
lr = 0.01
step_size = loss_slope * lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred1 = ((48.53 * X) + b).reshape(10)

plt.scatter(X, y)

plt.plot(
    X,
    reg.predict(X),
    color='red',
    label='OLS'
)

plt.plot(
    X,
    y_pred1,
    color='#00a65a',
    label='b = {}'.format(b)
)

plt.plot(
    X,
    y_pred,
    color='#A3E4D7',
    label='b = 0'
)

plt.legend()

plt.show()
```

```python
class GDRegressor:

    def __init__(self, learning_rate, epochs):

        self.m = 29.19
        self.b = -120

        self.lr = learning_rate
        self.epochs = epochs

    def fit(self, X, y):

        # calculate the b using GD
        for i in range(self.epochs):

            loss_slope = -2 * np.sum(
                y - self.m * X.ravel() - self.b
            )

            self.b = self.b - (
                self.lr * loss_slope
            )

            print(loss_slope, self.b)

        print(self.b)

```
