```python
import pandas as pd

import seaborn as sns
df = sns.load_dataset('titanic')
data = df[["survived","pclass","age","fare"]]
data
```

```python
data.isnull().sum()
```

```python
data['age'] = data['age'].fillna(data['age'].median())
```

```python
data.isnull().sum()
```

```python
import scipy.stats as stat
import pylab
import matplotlib.pyplot as plt
```

```python
#### If you want to check whether feature is gaussian or normal distributed
#### Q-Q plot

def plot_data(df, feature):
    plt.figure(figsize=(10,6))
    
    plt.subplot(1,2,1)
    df[feature].hist()
    
    plt.subplot(1,2,2)
    stat.probplot(df[feature], dist='norm', plot=pylab)
    
    plt.show()
```

```python
plot_data(data,'age')
```

```python
# Logarithmic Transformation

import numpy as np
data['age_log'] = np.log(data['age'])
plot_data(data, 'age_log')
```

```python
# Reciprocal Transformation

data['age_reciprocal'] = 1 / data.age
plot_data(data, 'age_reciprocal')
```

```python
##### Square Root Transformation

data['age_square'] = data.age**(1/2)
plot_data(data, 'age_square')
```

```python
#### Exponential Transformation

data['age_exponential'] = data.age**(1/1.2)
plot_data(data, 'age_exponential')
```

```python
import matplotlib.pyplot as plt
import scipy.stats as stat
import pylab

def allplot_data(df, feature):
    
    plt.figure(figsize=(20,10))
    
    # 1. Original Histogram
    plt.subplot(2,2,1)
    df[feature].hist(bins=30)
    plt.title(f'{feature} Histogram')
    
    
    # 2. Q-Q Plot
    plt.subplot(2,2,2)
    stat.probplot(df[feature].dropna(), dist='norm', plot=pylab)
    plt.title(f'{feature} Q-Q Plot')
    
    
    # 3. Box Plot
    plt.subplot(2,2,3)
    plt.boxplot(df[feature].dropna())
    plt.title(f'{feature} Box Plot')
    
    
    # 4. Density Plot
    plt.subplot(2,2,4)
    df[feature].plot(kind='density')
    plt.title(f'{feature} Density Plot')
    
    
    plt.tight_layout()
    plt.show()


# Calling Function
allplot_data(data, 'age')
```

```python

data['age_exponential'] = data.age**(1/1.2)
allplot_data(data, 'age_exponential')
```
