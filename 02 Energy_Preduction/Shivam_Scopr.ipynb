```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
# first check wheter we can apply the linner regration onthis data set using correlation method
new_data = data.iloc[:, 1:]
plt.figure(figsize=(20, 20))

sns.heatmap(
    new_data.corr(),
    annot=True,
    annot_kws={"size":12}   # Smaller annotation text
)

plt.tight_layout()
plt.show()
```

```python
x_new = new_data.iloc[:,:-1]
y_new = new_data["target"]
```

```python
sc = StandardScaler()
sc.fit(x_new)
x_new =pd.DataFrame(sc.transform(x_new),columns=x_new.columns)
x_new.head()
```

```python
x_train,x_test,y_train,y_test = train_test_split(x_new,y_new,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression, Lasso, Ridge
```

```python
lr = LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
lr_score = lr.score(x_test,y_test)*100
```

```python
lr.coef_
```

```python
plt.figure(figsize=(18,6))

plt.bar(x_new.columns, lr.coef_)

plt.xticks(rotation=45,fontsize = 12)
plt.tight_layout()
plt.title("Linear Regration")
plt.ylabel("coffecent")
plt.xlabel("name of feature")
plt.show()
```

```python
print(f"""MSE :{mean_squared_error(y_test,lr.predict(x_test))}
\nMAE: {mean_absolute_error(y_test,lr.predict(x_test))}
\nRMSE: {np.sqrt(mean_squared_error(y_test,lr.predict(x_test)))}""")
```

```python
la = Lasso(alpha=0.5)
la.fit(x_train,y_train)
```

```python
la_score = la.score(x_test,y_test)*100
```

```python
plt.figure(figsize=(18,5))

plt.bar(x_new.columns,la.coef_)
plt.xticks(rotation=45,fontsize = 12)
plt.title("Lasso")
plt.ylabel("coffecent")
plt.show()
```

```python
# X locations for the bars
x = np.arange(len(x_new.columns))
width = 0.4

plt.figure(figsize=(18,6))

# Linear Regression
plt.bar(x - width/2, lr.coef_, width=width, label='Linear Regression')

# Lasso
plt.bar(x + width/2, la.coef_, width=width, label='Lasso')

plt.xticks(x, x_new.columns, rotation=45, ha='right', fontsize=11)

plt.xlabel("Features")
plt.ylabel("Coefficient")
plt.title("Linear Regression vs Lasso Coefficients")

plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
print(f"""MSE :{mean_squared_error(y_test,la.predict(x_test))}
\nMAE: {mean_absolute_error(y_test,la.predict(x_test))}
\nRMSE: {np.sqrt(mean_squared_error(y_test,la.predict(x_test)))}""")
```

```python
# Ridge

ri = Ridge(alpha=10)
```

```python
ri.fit(x_train,y_train)
```

```python
ri_score = ri.score(x_test,y_test)*100
```

```python
print(f"LinearRegression : {lr_score} \nLasso : {la_score} \nRidge : {ri_score}")
```

```python
# X locations for the bars
x = np.arange(len(x_new.columns))
width = 0.40

plt.figure(figsize=(18,6))

# Linear Regression
plt.bar(x - width, lr.coef_, width=width, label='Linear Regression')

# Lasso
plt.bar(x, la.coef_, width=width, label='Lasso')

plt.bar(x + width, ri.coef_, width=width, label='Ridge')

plt.xticks(x, x_new.columns, rotation=45, ha='right', fontsize=11)

plt.xlabel("Features")
plt.ylabel("Coefficient")
plt.title("Linear Regression vs Lasso vs Ridge Coefficients")

plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
print(f"""MSE :{mean_squared_error(y_test,ri.predict(x_test))}
\nMAE: {mean_absolute_error(y_test,ri.predict(x_test))}
\nRMSE: {np.sqrt(mean_squared_error(y_test,ri.predict(x_test)))}""")
```
