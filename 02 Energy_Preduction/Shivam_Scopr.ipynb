```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
data = pd.read_csv("Energy_Dataset_ML.csv")
data.head()
```

```python
data["Time"] = pd.to_datetime(data["Time"])
data.info()
```

```python
plt.figure(figsize=(15, 10))

sns.boxplot(data=data.iloc[:, :-1], orient="h")

plt.title("Boxplot of All Features")
plt.xlabel("Values")
plt.show()
```

```python
features = data.iloc[:,1:]

# Create figure
plt.figure(figsize=(20, 20))

# Plot each feature
for i, col in enumerate(features.columns, 1):
    plt.subplot(5, 5, i)   # 5 rows × 5 columns = 25 plots
    sns.boxplot(y=features[col])
    plt.title(col)
    plt.ylabel("")

plt.tight_layout()
plt.show()
```

1. IQR Capping : Remove the value whihc is the out side of lower and upper boundy
 - Interquartile Range  : (IQR):IQR= Q3−Q1​
 - Lower Boundary, Upper Boundary​

2. Percentile Capping : Remove the value whihc is out side of this boundy
 - 1st Percentile , 99th Percentile
 - 5th Percentile , 95th Percentile

3. Winsorization : Replace the value whihc is out side of this boundy
 - 1st Percentile , 99th Percentile
 - 5th Percentile , 95th Percentile


```python
# Feature columns
X = data.iloc[:, 1:-1]
y = data["target"]
```

```python
# Figure size
plt.figure(figsize=(20, 20))

# Plot each feature vs target
for i, col in enumerate(X.columns, 1):
    plt.subplot(5, 5, i)
    sns.regplot(
        x=X[col],
        y=y,
        scatter_kws={"s": 20, "alpha": 0.6},
        line_kws={"color": "red"}
    )
    plt.title(f"{col} vs {y.name}")

plt.tight_layout()
plt.show()
```
