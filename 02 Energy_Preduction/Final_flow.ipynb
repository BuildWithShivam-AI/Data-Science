```python
# This is for assending if any time in data is up and down then it should reposition on that correct position
data = data.sort_values('Time')
```

```python
full_range = pd.date_range(
    start=data['Time'].min(),
    end=data['Time'].max(),
    freq='15min'
)

missing_times = full_range.difference(data['Time'])

print("Missing timestamps:", len(missing_times))
```

```python
for col in data.select_dtypes(include=['number']).columns:

    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((data[col] < lower) | (data[col] > upper)).sum()

    print(f"{col}: {outliers} Lower :{lower} Upper :{upper} min:{data[col].min()} max : {data[col].max()}")
```

```python
outlier_summary = []

for col in data.select_dtypes(include='number').columns:

    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((data[col] < lower) | (data[col] > upper)).sum()

    outlier_summary.append([
        col,
        outliers,
        (outliers/len(data))*100,
        lower,
        upper,
        data[col].min(),
        data[col].max(),
        Q1,
        Q3
    ])

outlier_df = pd.DataFrame(
    outlier_summary,
    columns=[
        'Feature',
        'Outlier_Count',
        'Outlier_%',
        'Lower_Limit',
        'Upper_Limit',
        'Min_Value',
        'Max_Value',
        'Q1',
        'Q3'
    ]
)

outlier_df
```
