```python
from nltk.stem import PorterStemmer
stemming = PorterStemmer()
```

```python
words = [
    "eating", "eats", "eaten",
    "writing", "writes", "written",
    "programming", "programs", "programmed",
    "running", "runs", "ran",
    "playing", "plays", "played",
    "walking", "walks", "walked",
    "reading", "reads", "read",
    "studying", "studies"
]
```

```python
for word in words:
    print(word+"--->"+stemming.stem(word))
```

```python
#Lancaster Stemming algorithm

from nltk.stem import LancasterStemmer
lancaster = LancasterStemmer()

for word in words:
    print(word+"--->"+lancaster.stem(word))
```

# RegexpStemmer class
NLTK has RegexpStemmer class with the help of which we can easily implement Regular Expression Stemmer algorithms. It basically takes a single regular expression and removes any prefix or suffix that matches the expression. Let us see an example

```python
from nltk.stem import RegexpStemmer
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$',min=4)
```

```python
for regword in words:
    print(regword+"--->"+reg_stemmer.stem(regword))
```

```python
#snowball Stemmer 
from nltk.stem import SnowballStemmer

snowballstemmer = SnowballStemmer('english',ignore_stopwords=False)
```

```python
for snowballword in words:
    print(snowballword+"--->"+lancaster.stem(snowballword))
```

# Wordnet Lemmatizer
- Lemmatization technique is like stemming. The output we will get after lemmatization is called ‘lemma’, which is a root word rather than root stem, the output of stemming. After lemmatization, we will be getting a valid word that means the same thing.

- NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us understand it with an example

```python
import nltk
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
for word in words:
    print(word+"----->"+lemmatizer.lemmatize(word))
```
