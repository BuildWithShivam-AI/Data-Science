```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("Titanic-Dataset.csv")
data.head()
```

```python
data["Age"] = data["Age"].fillna(data["Age"].mean(),inplace=True)
data.describe()
```

### Covariance 
- formula :  Data Present in range of ( - Infinity to + Infinity )





- Type of Covariance
  * Positive Covariance => X(increase) compart with Y(increase)
  * Negative Covariance => X(Decrease) compart with Y(increase)
  * Zero Covariance     => X(increase) compart with Y(random)

### Correlation 
- Formula : Data present in range of ( -1 to 1 )

```python
data = data.drop(columns=["PassengerId","Survived"],axis=0)
```

```python
data.info()
```

```python
data_corr = data.select_dtypes(["int64","float64"]).corr()
```

```python
data_cov = data.select_dtypes(["int64","float64"]).cov()
```

```python
sns.heatmap(data_corr,annot=True)
```

```python
sns.heatmap(data_cov,annot=True)
```
